```python
### Nested Dictionary
d = {"age":20 , "name": "hello"}
d.update({"name" : "arya mains"})
print(d)

l = [20,10]
print(l[0])

# nested dictionary
d= {"address": {"address":{"city":"kukas city" ,"district":"jaipur"},
                "address2":{"city" :"gopalpura" , "district":"arya main"}}}

print(d["address"]["address"]["city"])
print(d["address"]["address"]["district"])
print(d["address"]["address2"]["city"])
print(d["address"]["address2"]["district"])

# same but different method
dict = {"address":{"address1":{"city":"kukas","district":"jaipur"},
                   "address2":{"city":"andheri","district":"mumbai"}}}
for i in dict["address"]["address1"]:
    print(i)
    print(dict["address"]["address1"][i])
for i in dict["address"]["address2"]:
    print(i)
    print(dict["address"]["address2"][i])

# agar data agar large ho to list ke bajaye dict ka use karenge
#  next question for nested dictionary




```
